```python

import numpy as np
import rasterio
import gc

from rasterio.windows import Window, from_bounds
from rasterio.features import geometry_mask


def sample_url_worker_polygon(url, polygons, stat="mean"):
    """
    Memory-efficient polygon sampling.
    Reads only the required raster windows and individual
    centroid pixels when fallback sampling is necessary.
    """

    results = []

    def read_centroid(src, poly):
        row, col = src.index(
            poly.centroid.x,
            poly.centroid.y
        )

        if not (0 <= row < src.height and
                0 <= col < src.width):
            return np.nan

        pixel = src.read(
            1,
            window=Window(col, row, 1, 1),
            masked=True
        )

        if pixel.size == 0 or np.ma.is_masked(pixel[0, 0]):
            return np.nan

        value = float(pixel[0, 0])

        return value if np.isfinite(value) else np.nan

    try:
        with rasterio.open(url) as src:

            res_x, res_y = src.res

            for poly in polygons:

                minx, miny, maxx, maxy = poly.bounds

                # Small polygons: centroid sampling
                if ((maxx - minx) < res_x or
                        (maxy - miny) < res_y):

                    results.append(read_centroid(src, poly))
                    continue

                # Calculate raster window
                window = from_bounds(
                    minx, miny, maxx, maxy,
                    transform=src.transform
                )

                window = window.round_offsets().round_lengths()

                if window.width <= 0 or window.height <= 0:
                    results.append(read_centroid(src, poly))
                    continue

                # Read only the required raster window
                data = src.read(
                    1,
                    window=window,
                    masked=True
                )

                transform = src.window_transform(window)

                mask = geometry_mask(
                    [poly],
                    transform=transform,
                    invert=True,
                    out_shape=data.shape
                )

                vals = data[mask]

                # Remove masked and invalid pixels
                vals = np.ma.asarray(vals).compressed()
                vals = vals[np.isfinite(vals)]

                # Fallback if no valid pixels
                if vals.size == 0:
                    results.append(read_centroid(src, poly))
                    del data, mask, vals
                    continue

                # Calculate requested statistic
                if stat == "mean":
                    value = np.mean(vals)

                elif stat == "sum":
                    value = np.sum(vals)

                elif stat == "min":
                    value = np.min(vals)

                elif stat == "max":
                    value = np.max(vals)

                elif stat == "median":
                    value = np.median(vals)

                else:
                    raise ValueError(
                        f"Unsupported statistic: {stat}"
                    )

                results.append(float(value))

                del data, mask, vals

                # Release temporary memory periodically
                if len(results) % 25 == 0:
                    gc.collect()

        return results

    except Exception as e:
        print(f"Sampling failed for {url}: {e}")
        return [np.nan] * len(polygons)


# Replace the function inside the imported module
wdl.sample_url_worker_polygon = sample_url_worker_polygon

print("Memory-efficient raster worker installed!")
```

```python
import pandas as pd
import os

output_dir = "/content/WaPOR_Output/Iraq/csvs"

datasets = {
    "PCP": "AgERA5-PCP-M_mm_clean.csv",
    "RET": "AgERA5-RET-M_mm_clean.csv"
}

for product, filename in datasets.items():

    path = os.path.join(output_dir, filename)
    df = pd.read_csv(path, parse_dates=["Date"])

    # Select governorate columns
    governorate_data = df.drop(columns=["Date"])

    # Calculate descriptive statistics
    stats = governorate_data.describe().T

    stats = stats[["mean", "std", "min", "50%", "max"]]
    stats = stats.rename(columns={
        "mean": "Mean",
        "std": "Std_Dev",
        "min": "Minimum",
        "50%": "Median",
        "max": "Maximum"
    })

    # Save statistics
    stats_path = os.path.join(
        output_dir,
        f"AgERA5_{product}_Descriptive_Statistics.csv"
    )

    stats.to_csv(stats_path)

    print("\n" + "=" * 65)
    print(f"{product} DESCRIPTIVE STATISTICS (mm/month)")
    print("=" * 65)

    display(stats.round(2))

    print("Saved:", stats_path)
```
